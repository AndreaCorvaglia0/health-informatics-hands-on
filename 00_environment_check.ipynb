## 3. Dataset

```python
from pathlib import Path

# On Kaggle the datasets added with "Add Input" are under /kaggle/input, in a subfolder
# (datasets/<owner>/). We look for the course dataset by its name.
DATA_DIR = next(Path("/kaggle/input").rglob("luna16-lung-slices"), Path("/kaggle/input/luna16-lung-slices"))

if DATA_DIR.exists():
    n_images = len(list((DATA_DIR / "images").glob("*.nii.gz")))
    n_labels = len(list((DATA_DIR / "labels").glob("*.nii.gz")))
    print(f"{n_images} images and {n_labels} masks found in {DATA_DIR}")
    print("Dataset: OK")
else:
    print(f"{DATA_DIR} not found.")
    print("Right panel -> Input -> Add Input -> search the course dataset -> click +. Then run this cell again.")
    print("Datasets attached:", [str(p) for p in Path("/kaggle/input").glob("*/*/*")])
```

## 4. A first image

If the three checks above print `OK`, this cell shows one CT slice with its lung mask.

```python
import matplotlib.pyplot as plt
import nibabel as nib

image = nib.load(DATA_DIR / "images" / "case_0001.nii.gz").get_fdata()
mask = nib.load(DATA_DIR / "labels" / "case_0001.nii.gz").get_fdata()

plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.title("CT slice")
plt.imshow(image.T, cmap="gray", vmin=-1000, vmax=400)
plt.axis("off")
plt.subplot(1, 2, 2)
plt.title("Lung mask")
plt.imshow(mask.T, cmap="gray")
plt.axis("off")
plt.show()
```
